```python
import vertexai
from vertexai.generative_models import GenerativeModel, Part

PROJECT_ID = "qwiklabs-gcp-03-42d2ec04c144"
LOCATION = "us-central1"

def classify_question_with_gemini(question: str) -> str:
    """
    Classifies a user question into one of predefined categories using Gemini.

    Args:
        project_id: Your Google Cloud project ID.
        location: The Google Cloud region where Vertex AI is enabled (e.g., "us-central1").
        question: The user question to classify.

    Returns:
        The classified category (Employment, General Information, Emergency Services,
        or Tax Related), or "Unclassified" if Gemini cannot determine a category.
    """
    vertexai.init(project=PROJECT_ID, location=LOCATION)
    model = GenerativeModel("gemini-2.5-flash")

    prompt = f"""
    Classify the following question into one of these categories:
    Employment, General Information, Emergency Services, Tax Related.

    Examples:
    Question: I need to report a fire.
    Category: Emergency Services

    Question: What are the hours of operation for the public library?
    Category: General Information

    Question: How do I apply for a job?
    Category: Employment

    Question: I received a notice about my property taxes.
    Category: Tax Related

    Question: What is the weather forecast for tomorrow?
    Category: General Information

    Question: My car broke down on the highway.
    Category: Emergency Services

    Question: I need information on employee benefits.
    Category: Employment

    Question: How do I file my annual income tax return?
    Category: Tax Related

    Question: {question}
    Category:
    """

    try:
        response = model.generate_content(prompt)
        # Gemini often returns the category directly, but sometimes includes
        # extra text. We'll try to extract just the category.
        classified_category = response.text.strip()

        # Simple validation to ensure the response is one of our categories
        valid_categories = ["Employment", "General Information", "Emergency Services", "Tax Related"]
        if classified_category in valid_categories:
            return classified_category
        else:
            # If Gemini gives a different response, we can try to be more robust
            # or return "Unclassified". For now, we'll try to find a keyword.
            for category in valid_categories:
                if category.lower() in classified_category.lower():
                    return category
            return "Unclassified" # Default if it doesn't match
    except Exception as e:
        print(f"An error occurred: {e}")
        return "Error"

# --- Example Usage ---
if __name__ == "__main__":
    # Replace with your actual project ID and desired location
    my_project_id = "your-gcp-project-id" # e.g., "qwiklabs-gcp-03-42d2ec04c144"
    my_location = "us-central1"

    test_questions = [
        "I need help, there's an intruder in my house!",
        "Where can I find information about public transport schedules?",
        "What are the requirements for a software engineer role?",
        "When is the deadline for submitting tax forms?",
        "Tell me a joke.",
        "I need to report a car accident."
    ]

    for q in test_questions:
        category = classify_question_with_gemini(q)
        print(f"Question: '{q}'\nClassified as: {category}\n")

```

```python
# prompt: Create a python function that generates social media posts for government announcements like
# weather emergencies, holidays, school closings, etc.

import vertexai
from vertexai.generative_models import GenerativeModel, Part

PROJECT_ID = "qwiklabs-gcp-03-42d2ec04c144"
LOCATION = "us-central1"

def generate_social_media_post(announcement_type: str, details: str) -> str:
    """
    Generates a social media post for government announcements using Gemini.

    Args:
        announcement_type: The type of announcement (e.g., "Weather Emergency", "Holiday", "School Closing").
        details: Specific details about the announcement.

    Returns:
        A generated social media post.
    """
    vertexai.init(project=PROJECT_ID, location=LOCATION)
    model = GenerativeModel("gemini-2.5-flash")

    prompt = f"""
    Generate a concise and informative social media post for a government announcement.
    Include relevant hashtags.

    Announcement Type: {announcement_type}
    Details: {details}

    Social Media Post:
    """

    try:
        response = model.generate_content(prompt)
        return response.text.strip()
    except Exception as e:
        print(f"An error occurred: {e}")
        return "Error generating post."

# --- Example Usage ---
if __name__ == "__main__":
    # Example 1: Weather Emergency
    weather_announcement = {
        "type": "Weather Emergency",
        "details": "A severe thunderstorm warning is in effect for the metropolitan area. Residents are advised to seek shelter indoors and avoid travel."
    }
    post1 = generate_social_media_post(weather_announcement["type"], weather_announcement["details"])
    print(f"--- Weather Emergency Post ---")
    print(post1)
    print("\n")

    # Example 2: Holiday Announcement
    holiday_announcement = {
        "type": "Holiday",
        "details": "City Hall and all municipal offices will be closed on Monday, January 15th, in observance of Martin Luther King Jr. Day."
    }
    post2 = generate_social_media_post(holiday_announcement["type"], holiday_announcement["details"])
    print(f"--- Holiday Post ---")
    print(post2)
    print("\n")

    # Example 3: School Closing
    school_closing_announcement = {
        "type": "School Closing",
        "details": "All public schools in the district will be closed tomorrow, February 10th, due to icy road conditions. Please check school websites for updates."
    }
    post3 = generate_social_media_post(school_closing_announcement["type"], school_closing_announcement["details"])
    print(f"--- School Closing Post ---")
    print(post3)
    print("\n")
```

```python
import pytest

# Assuming generate_social_media_post is defined in the current environment

def test_generate_weather_emergency_post():
    announcement_type = "Weather Emergency"
    details = "A severe thunderstorm warning is in effect for the metropolitan area. Residents are advised to seek shelter indoors and avoid travel."
    post = generate_social_media_post(announcement_type, details)
    assert "thunderstorm" in post.lower()
    assert "shelter" in post.lower()
    assert "#weatheralert" in post.lower() or "#emergency" in post.lower()
    assert len(post) > 50 # Ensure a reasonably long post

def test_generate_holiday_announcement_post():
    announcement_type = "Holiday"
    details = "City Hall and all municipal offices will be closed on Monday, January 15th, in observance of Martin Luther King Jr. Day."
    post = generate_social_media_post(announcement_type, details)
    assert "city hall" in post.lower()
    assert "closed" in post.lower()
    assert "martin luther king jr. day" in post.lower() or "#mlkday" in post.lower()
    assert len(post) > 50

def test_generate_school_closing_post():
    announcement_type = "School Closing"
    details = "All public schools in the district will be closed tomorrow, February 10th, due to icy road conditions."
    post = generate_social_media_post(announcement_type, details)
    assert "schools" in post.lower()
    assert "closed" in post.lower()
    assert "icy road conditions" in post.lower() or "#schoolclosure" in post.lower()
    assert len(post) > 50

def test_generate_generic_announcement_post():
    announcement_type = "Public Health Advisory"
    details = "Wash your hands frequently and maintain social distancing."
    post = generate_social_media_post(announcement_type, details)
    assert "wash your hands" in post.lower() or "social distancing" in post.lower()
    assert len(post) > 30
```
